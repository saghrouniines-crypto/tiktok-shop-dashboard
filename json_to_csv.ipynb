```python
import pandas as pd

def get_matching_variant_details(row):
    # Combine title and description text for matching
    text_to_search = f"{str(row['title'])} {str(row.get('description', ''))}".lower()
    variants = row['variants']
    
    if not isinstance(variants, list):
        return None
    
    for variant in variants:
        if not isinstance(variant, dict):
            continue
            
        # Check property values (e.g. 'Galaxy A55', 'iPhone 16 Pro Max')
        properties = variant.get('properties', [])
        for prop in properties:
            val = str(prop.get('value', '')).lower()
            # If the specific model or property value is explicitly named in the title/desc
            if val and len(val) > 2 and val in text_to_search:
                return {
                    'matched_variant_id': variant.get('variantId'),
                    'matched_stock': variant.get('availableQuantity'),
                    'matched_price': variant.get('salePrice', {}).get('value')
                }
                
    return None

# Apply to your DataFrame
matched_data = df.apply(get_matching_variant_details, axis=1)

# Extract into clean columns on your 1-row-per-product DataFrame
df['matched_variant_id'] = matched_data.apply(lambda x: x['matched_variant_id'] if x else None)
df['matched_stock'] = matched_data.apply(lambda x: x['matched_stock'] if x else None)
df['matched_price'] = matched_data.apply(lambda x: x['matched_price'] if x else None)

print(df[['productId', 'title', 'matched_variant_id', 'matched_price', 'matched_stock']].head())
print(df)
```
